```python
import pickle as pk # Imports the Pickle library, which is a standard tool for saving (serializing) and loading (de-serializing) objects.


loaded_model= pk.load(open('finalized_model_slr.sav','rb')) # opens the file named finalized_model_slr.sav in Read Binary mode.

result = loaded_model.predict([[100]]) # This takes the value 100 and passes it into the model to get a result. The double brackets [[ ]] are used because most models expect a 2D array (a list of rows) as input.
```
